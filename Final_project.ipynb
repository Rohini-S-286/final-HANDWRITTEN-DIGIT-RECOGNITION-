```python
# ============================================================
# HANDWRITTEN DIGIT RECOGNITION AI
# MNIST + CNN + OpenCV preprocessing + Gradio
# ============================================================

import os
import zipfile
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import tensorflow as tf
import gradio as gr

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.metrics import confusion_matrix, classification_report


# ============================================================
# 1. CHECK TENSORFLOW / GPU
# ============================================================

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))


# ============================================================
# 2. LOAD MNIST DATASET
# ============================================================

(X_train, y_train), (X_test, y_test) = mnist.load_data()

print("Training images:", X_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", X_test.shape)
print("Testing labels:", y_test.shape)


# ============================================================
# 3. DISPLAY SAMPLE TRAINING IMAGES
# ============================================================

plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(f"Digit: {y_train[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 4. NORMALIZE DATA
# ============================================================

X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0


# ============================================================
# 5. ADD CHANNEL DIMENSION
# ============================================================

X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
print("Minimum pixel value:", X_train.min())
print("Maximum pixel value:", X_train.max())


# ============================================================
# 6. DATA AUGMENTATION
# ============================================================

data_augmentation = tf.keras.Sequential(
    [
        layers.RandomRotation(0.08),
        layers.RandomZoom(0.10),
        layers.RandomTranslation(0.10, 0.10),
    ],
    name="data_augmentation",
)

print("Data augmentation created successfully!")


# Display augmented examples
plt.figure(figsize=(10, 5))

for i in range(10):
    augmented = data_augmentation(
        X_train[i : i + 1],
        training=True,
    )

    plt.subplot(2, 5, i + 1)
    plt.imshow(augmented[0, :, :, 0], cmap="gray")
    plt.title(f"Digit: {y_train[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 7. BUILD CNN MODEL
# ============================================================

model = models.Sequential(
    [
        layers.Input(shape=(28, 28, 1)),

        data_augmentation,

        layers.Conv2D(32, (3, 3), activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(64, (3, 3), activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(128, (3, 3), activation="relu"),
        layers.BatchNormalization(),

        layers.Flatten(),

        layers.Dense(128, activation="relu"),
        layers.Dropout(0.4),

        layers.Dense(10, activation="softmax"),
    ],
    name="mnist_cnn",
)

model.summary()


# ============================================================
# 8. COMPILE MODEL
# ============================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

print("Model compiled successfully!")


# ============================================================
# 9. CALLBACKS
# ============================================================

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True,
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=0.00001,
)

print("Callbacks ready!")


# ============================================================
# 10. TRAIN MODEL
# ============================================================

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.1,
    callbacks=[early_stop, reduce_lr],
    verbose=1,
)


# ============================================================
# 11. EVALUATE MODEL
# ============================================================

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0,
)

print("Test Accuracy:", round(test_accuracy * 100, 2), "%")
print("Test Loss:", round(test_loss, 4))


# ============================================================
# 12. CONFUSION MATRIX + CLASSIFICATION REPORT
# ============================================================

probabilities = model.predict(
    X_test,
    batch_size=128,
    verbose=1,
)

y_pred = np.argmax(probabilities, axis=1)

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
)

plt.title("Confusion Matrix - Handwritten Digit Recognition")
plt.xlabel("Predicted Digit")
plt.ylabel("Actual Digit")
plt.tight_layout()
plt.show()

print("\nClassification Report:")
print(classification_report(y_test, y_pred))


# ============================================================
# 13. SAVE MODEL
# ============================================================

MODEL_PATH = "handwritten_digit_recognition.keras"

model.save(MODEL_PATH)

print(f"Model saved successfully: {MODEL_PATH}")


# ============================================================
# 14. SHOW RANDOM TEST PREDICTIONS
# ============================================================

indices = np.random.choice(
    len(X_test),
    10,
    replace=False,
)

plt.figure(figsize=(15, 4))

for i, idx in enumerate(indices):

    prediction = int(
        np.argmax(probabilities[idx])
    )

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        X_test[idx].squeeze(),
        cmap="gray",
    )

    plt.title(
        f"Actual: {y_test[idx]}\n"
        f"Predicted: {prediction}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 15. REAL IMAGE PREPROCESSING
# ============================================================

def preprocess_handwriting(image):
    """
    Convert a real handwritten digit image into an MNIST-like
    28x28 image with a black background and white digit.
    """

    if image is None:
        return np.zeros(
            (28, 28),
            dtype=np.float32,
        )

    image = np.asarray(image)

    # Convert to grayscale
    if image.ndim == 3:

        if image.shape[2] == 4:

            gray = cv2.cvtColor(
                image,
                cv2.COLOR_RGBA2GRAY,
            )

        else:

            gray = cv2.cvtColor(
                image,
                cv2.COLOR_RGB2GRAY,
            )

    elif image.ndim == 2:

        gray = image.copy()

    else:

        return np.zeros(
            (28, 28),
            dtype=np.float32,
        )

    # Make sure OpenCV receives uint8 data
    if gray.dtype != np.uint8:

        gray = cv2.normalize(
            gray,
            None,
            0,
            255,
            cv2.NORM_MINMAX,
        ).astype(np.uint8)

    # Reduce small camera/pencil noise
    gray = cv2.GaussianBlur(
        gray,
        (3, 3),
        0,
    )

    # Adaptive threshold
    binary = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY_INV,
        31,
        9,
    )

    # Remove long horizontal notebook lines
    kernel_width = max(
        25,
        gray.shape[1] // 8,
    )

    horizontal_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT,
        (kernel_width, 1),
    )

    horizontal_lines = cv2.morphologyEx(
        binary,
        cv2.MORPH_OPEN,
        horizontal_kernel,
    )

    clean = cv2.subtract(
        binary,
        horizontal_lines,
    )

    # Remove tiny noise
    clean = cv2.morphologyEx(
        clean,
        cv2.MORPH_OPEN,
        np.ones((2, 2), np.uint8),
    )

    # Find connected components
    num_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            clean,
            8,
        )
    )

    components = []

    for i in range(1, num_labels):

        x, y, w, h, area = stats[i]

        if area > 30 and h > 10 and w > 3:

            components.append(
                (x, y, w, h, area)
            )

    if not components:

        return np.zeros(
            (28, 28),
            dtype=np.float32,
        )

    # Bounding box around handwritten digit
    x1 = min(
        component[0]
        for component in components
    )

    y1 = min(
        component[1]
        for component in components
    )

    x2 = max(
        component[0] + component[2]
        for component in components
    )

    y2 = max(
        component[1] + component[3]
        for component in components
    )

    # Add padding
    pad = 15

    x1 = max(
        0,
        x1 - pad,
    )

    y1 = max(
        0,
        y1 - pad,
    )

    x2 = min(
        clean.shape[1],
        x2 + pad,
    )

    y2 = min(
        clean.shape[0],
        y2 + pad,
    )

    digit = clean[
        y1:y2,
        x1:x2,
    ]

    if digit.size == 0:

        return np.zeros(
            (28, 28),
            dtype=np.float32,
        )

    # Make square while preserving aspect ratio
    h, w = digit.shape

    size = max(h, w)

    square = np.zeros(
        (size, size),
        dtype=np.uint8,
    )

    x_offset = (size - w) // 2
    y_offset = (size - h) // 2

    square[
        y_offset:y_offset + h,
        x_offset:x_offset + w,
    ] = digit

    # Resize to MNIST size
    digit28 = cv2.resize(
        square,
        (28, 28),
        interpolation=cv2.INTER_AREA,
    )

    # Normalize
    return digit28.astype(
        "float32"
    ) / 255.0


# ============================================================
# 16. PREDICTION FUNCTION
# ============================================================

def recognize_digit(image):
    """
    Predict one handwritten digit and return:
    prediction, confidence, processed image.
    """

    if image is None:

        return (
            "Upload an image",
            "0.00%",
            None,
        )

    processed = preprocess_handwriting(
        image
    )

    probabilities = model.predict(
        processed.reshape(
            1,
            28,
            28,
            1,
        ),
        verbose=0,
    )[0]

    predicted = int(
        np.argmax(probabilities)
    )

    confidence = float(
        np.max(probabilities) * 100
    )

    return (
        f"Predicted Digit: {predicted}",
        f"{confidence:.2f}%",
        processed,
    )


# ============================================================
# 17. FINAL MODEL SANITY CHECK
# ============================================================

idx = 100

image = X_test[idx]

actual = int(
    y_test[idx]
)

prob = model.predict(
    image.reshape(
        1,
        28,
        28,
        1,
    ),
    verbose=0,
)[0]

predicted = int(
    np.argmax(prob)
)

plt.figure(figsize=(3, 3))

plt.imshow(
    image.squeeze(),
    cmap="gray",
)

plt.title(
    f"Actual: {actual} | "
    f"Predicted: {predicted}"
)

plt.axis("off")
plt.show()

print("Actual digit:", actual)
print("Predicted digit:", predicted)

print(
    "Confidence:",
    f"{prob[predicted] * 100:.2f}%",
)


# ============================================================
# 18. OPTIONAL: EVALUATE A ZIP OF LABELED IMAGES
# ============================================================

def extract_zip(
    zip_filename,
    extract_folder="handwritten_images",
):
    """
    Extract a ZIP file safely for optional
    external-image testing.
    """

    if not zipfile.is_zipfile(
        zip_filename
    ):

        raise ValueError(
            "The uploaded file is not a valid ZIP file."
        )

    os.makedirs(
        extract_folder,
        exist_ok=True,
    )

    with zipfile.ZipFile(
        zip_filename,
        "r",
    ) as zip_ref:

        zip_ref.extractall(
            extract_folder
        )

    print(
        "ZIP extracted successfully!"
    )

    print(
        "Folder:",
        extract_folder,
    )

    return extract_folder


# ============================================================
# 19. GRADIO APPLICATION
# ============================================================

custom_css = """

/* Main website */
.gradio-container {
    max-width: 1400px !important;
    margin: auto !important;
    padding: 30px !important;
}

/* Main heading */
h1 {
    font-size: 42px !important;
    text-align: center !important;
}

h2 {
    font-size: 30px !important;
}

/* Make image containers large */
.large-image {
    min-height: 500px !important;
}

/* Upload image area */
.large-upload {
    min-height: 500px !important;
}

/* Buttons */
button {
    font-size: 22px !important;
    padding: 18px 30px !important;
    min-height: 65px !important;
}

/* Textboxes */
textarea,
input {
    font-size: 22px !important;
}

/* Prediction and confidence */
.result-box textarea {
    font-size: 28px !important;
    font-weight: bold !important;
}

/* Markdown text */
.markdown-text {
    font-size: 20px !important;
}

/* Columns */
.gradio-row {
    gap: 30px !important;
}

"""


with gr.Blocks(
    title="Handwritten Digit Recognition AI",
    css=custom_css,
) as demo:

    # ========================================================
    # WEBSITE HEADER
    # ========================================================

    gr.Markdown(
        """
        # 🔢 AI Handwritten Digit Recognition

        ### CNN-based handwritten digit classification

        Upload a clear handwritten digit from **0–9**
        and let the AI recognize it.
        """
    )

    # ========================================================
    # MAIN AREA
    # ========================================================

    with gr.Row():

        # ====================================================
        # LEFT SIDE
        # ====================================================

        with gr.Column(
            scale=1,
            min_width=500,
        ):

            input_image = gr.Image(
                type="numpy",
                label="Upload Handwritten Digit",
                height=500,
                width=500,
                elem_classes="large-upload",
            )

            recognize_button = gr.Button(
                "🔍 Recognize Digit",
                variant="primary",
                size="lg",
            )

        # ====================================================
        # RIGHT SIDE
        # ====================================================

        with gr.Column(
            scale=1,
            min_width=500,
        ):

            prediction = gr.Textbox(
                label="Prediction",
                placeholder="Prediction will appear here",
                elem_classes="result-box",
                lines=2,
            )

            confidence = gr.Textbox(
                label="Confidence",
                placeholder="Confidence will appear here",
                elem_classes="result-box",
                lines=2,
            )

            processed = gr.Image(
                label="What the AI Actually Sees",
                height=500,
                width=500,
                elem_classes="large-image",
            )

    # ========================================================
    # BUTTON ACTION
    # ========================================================

    recognize_button.click(
        fn=recognize_digit,
        inputs=input_image,
        outputs=[
            prediction,
            confidence,
            processed,
        ],
    )

    # ========================================================
    # INFORMATION SECTION
    # ========================================================

    gr.Markdown(

    )


# ============================================================
# 20. LAUNCH GRADIO
# ============================================================

print("Launching Gradio application...")

demo.launch(
    share=True,
)
```
